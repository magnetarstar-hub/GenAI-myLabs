```python
#Generation d'images (text to image)
import torch
from diffusers import StableDiffusionPipeline
import os
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
os.makedirs("out", exist_ok=True)
```

```python
#Utilisation du GPU si disponible pour accelerer la generation

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Chargement du modèle sur {device}...")

pipe = StableDiffusionPipeline.from_pretrained(
    "stable-diffusion-v1-5/stable-diffusion-v1-5",
    torch_dtype=torch.float32,
    safety_checker=None,
    requires_safety_checker=False,
)
```

```python
if device == "cuda":
    pipe.enable_sequential_cpu_offload()   # replaces pipe.to(device)
    pipe.enable_attention_slicing()
else:
    pipe = pipe.to(device)
prompt = "A futuristic city with flying cars at sunset, cyberpunk style"
print("Génération de l'image en cours...")
image = pipe(prompt, num_inference_steps=25, height=512, width=512).images[0]
image.save("out/generated_image.png")
print("Image sauvegardée : out/generated_image.png")
```
