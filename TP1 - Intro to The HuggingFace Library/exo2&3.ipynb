```python
from transformers import pipeline
```

```python
#Initialisation du pipeline de generation
generator = pipeline('text-generation', model='gpt2')
prompt = "In the future, artificial intelligence will"
```

```python
#Generation standard
print("--- Génération Standard ---")
result_std = generator(prompt, max_length=50, num_return_sequences=1, truncation=True)
print(result_std[0]['generated_text'])
```

```python
# Temperature basse (Deterministe)
print("\n--- Temperature Basse (0.2) ---")
result_low = generator(prompt, max_length=50, temperature=0.2, do_sample=True, truncation=True)
print(result_low[0]['generated_text'])
```

```python
print("\n---Temperature Haute (0.9) ---")
result_high = generator(prompt, max_length=50, temperature=0.9, truncation=True, do_sample=True)
print(result_high[0]['generated_text'])
```
